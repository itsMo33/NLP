# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False)

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

text = "This NLP lab is easy and interesting."

result = classifier(text)
print(result)
````

**Why:**

- **Machine translation** and **summarization** take an input text and produce a *new* text from it, so they need an encoder to read the input and a decoder to write the output.
- **Open-ended text generation** is just predicting the next token again and again, which is exactly what a decoder-only model like GPT does.


## Task 5 — Short Reflection

Answer in 3–5 sentences:

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**


**Answer:**

A chatbot like ChatGPT has to write its reply word by word, and this is exactly how a decoder-only model works, because it is trained to predict the next token based on all the tokens before it. The conversation (the user's message and the previous replies) is given to the model as one long text, and the model simply continues it with an answer. It does not need a separate encoder for the input, since the decoder can read the whole conversation as context through self-attention. Also, decoder-only models are easy to train on huge amounts of text from the internet, which is why they learn a lot of general knowledge and can answer many different kinds of questions.

### Question

Why do you sometimes get different outputs even when the prompt is the same?

**Hint:** Sampling introduces randomness when choosing the next token.


````python
# Same prompt, with set_seed(42)
set_seed(42)

prompt = "Natural language processing is"

for i in range(3):
    output = generator(prompt, max_new_tokens=30, do_sample=True)
    print(f"Run {i + 1}: {output[0]['generated_text']}")
    print()
````

````python
# Same prompt, without set_seed(42)
prompt = "Natural language processing is"

for i in range(3):
    output = generator(prompt, max_new_tokens=30, do_sample=True)
    print(f"Run {i + 1}: {output[0]['generated_text']}")
    print()
````

**Answer:**

When `do_sample=True`, the model does not always pick the word with the highest probability. Instead, it picks the next token randomly based on the probabilities, so a word with 30% probability can still be chosen sometimes. Because this happens at every step, one different word at the start changes everything that comes after it, and that is why we get a different text each time even with the same prompt. `set_seed(42)` fixes the random number generator, so if we run the cell again from the same seed we get the same results. In Part A we used `do_sample=False` (greedy decoding), which always picks the most likely token, so the output was the same every time.
